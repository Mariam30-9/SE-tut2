```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel('credit-card_customers.xlsx')
df = df.drop_duplicates(subset='Customer_Number', keep='first')
df = df[~(df['Attrition_Flag'].isna() | (df['Attrition_Flag'] == 'Unknown'))]
df['Churned'] = (df['Attrition_Flag'] == 'Attrited Customer').astype(int)

df['Gender'] = df['Gender'].fillna('Unknown')
df['Card_Category'] = df['Card_Category'].fillna('Missing')
df['Marital_Status'] = df['Marital_Status'].fillna('Unknown')
df['Income_Category'] = df['Income_Category'].fillna('Unknown')
df['Education_Level'] = df['Education_Level'].fillna('Unknown')

cat_vars = {
    'Gender': ['F', 'M', 'Unknown'],
    'Marital_Status': ['Married', 'Single', 'Divorced', 'Unknown'],
    'Education_Level': ['Uneducated', 'High School', 'College', 'Graduate',
                          'Post-Graduate', 'Doctorate', 'Unknown'],
    'Income_Category': ['Less than $40K', '$40K - $60K', '$60K - $80K',
                          '$80K - $120K', '$120K +', 'Unknown'],
    'Card_Category': ['Blue', 'Silver', 'Gold', 'Platinum', 'Missing']
}

overall_rate = df['Churned'].mean() * 100

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

for i, (col, order) in enumerate(cat_vars.items()):
    ax = axes[i]
    summary = df.groupby(col)['Churned'].agg(['mean', 'count']).reindex(order)
    summary['pct'] = summary['mean'] * 100
    sns.barplot(x=summary.index, y=summary['pct'], hue=summary.index,
                legend=False, palette='rocket', order=order, ax=ax)
    ax.axhline(overall_rate, color='gray', linestyle='--', linewidth=1.2)
    for j, v in enumerate(summary['pct']):
        ax.text(j, v + 0.8, f'{v:.1f}%', ha='center', fontsize=8)
    ax.set_title(col, fontsize=12)
    ax.set_xlabel('')
    ax.set_ylabel('Churn Rate (%)')
    ax.set_ylim(0, 30)
    ax.tick_params(axis='x', rotation=35)

axes[-1].axis('off')
fig.suptitle('Churn Rate Across All Categorical Variables (dashed line = overall 16.7%)',
             fontsize=15, y=1.0)
plt.tight_layout()
plt.savefig('churn_rate_all_categoricals.png', dpi=200, bbox_inches='tight')
plt.show()
```
